```python
import os
import sys
import json
import pandas as pd

total_df = pd.DataFrame()
for file_name in os.listdir('./'):
    if not file_name.endswith('.json'):
        continue
    with open(file_name, 'r') as f:
        data = json.load(f)
    print(file_name)
    # ensure we have a DataFrame row with required fields
    if isinstance(data, dict):
        print
        record = {
            'mode': data.get('mode'),
            'coverage_area_num': data.get('coverage_area_num'),
            'operational_node_num': data.get('operational_node_num'),
            'reconstructed_total_time': data.get('reconstructed_total_time'),
        }
        data_df = pd.DataFrame([record])
    else:
        # e.g., list of dicts
        data_df = pd.DataFrame(data)[['mode', 'coverage_area_num', 'operational_node_num', 'reconstructed_total_time']]
    # print(data)
    total_df = pd.concat([total_df, data_df], ignore_index=True)
total_df.to_csv('results.csv', index=False)
```
