```python
import json
import csv
import pandas as pd
import numpy as np
```

```python
total_results = pd.DataFrame()
for mode in ['C', 'RC', 'R']:
    for coverage_area_num in [6,9,12]:
        for operational_node_num in [10,20,30]:
            file_path = f'./{coverage_area_num}/result_{mode}_{coverage_area_num}_{operational_node_num}.json'
            with open(file_path, 'r') as f:
                result = json.load(f)
            # 处理结果
            # print(result)
            result = {
                'global_optimal_time': result['global_optimal_time'],
                # 'total_local_search_weight': result['total_local_search_weight'],
                # 'shaking_method_first_num': result['shaking_method_first_num'],
                'init_time': result['init_time'],
                'improvement': result['improvement'],
                'iter_num': result['iter_num'],
                'time_elapsed': result['time_elapsed'],
                'global_optimal_time_var': result['global_optimal_time_var']
                }
            # Wrap dict in a list so pandas creates one row instead of raising ValueError for scalars
            df = pd.DataFrame([result])
            # df = df[['global_optimal_time', 'total_local_search_weight', 'shaking_method_first_num', 'init_time', 'improvement', 'iter_num', 'time_elapsed', 'global_optimal_time_var']]
            df['mode'] = mode
            df['coverage_area_num'] = coverage_area_num
            df['operational_node_num'] = operational_node_num
            total_results = pd.concat([total_results, df], ignore_index=True)
# 保存结果到CSV文件
total_results.to_csv('collected_results.csv', index=False)
```

```python
total_results = pd.DataFrame()
local_search_oeprator_weight_list = []
shaking_operator_first_num_list = []
for mode in ['C', 'RC', 'R']:
    for coverage_area_num in [6,9,12]:
        for operational_node_num in [10,20,30]:
            file_path = f'./{coverage_area_num}/result_{mode}_{coverage_area_num}_{operational_node_num}.json'
            with open(file_path, 'r') as f:
                result = json.load(f)
            local_search_oeprator_weight = result['total_local_search_weight']
            shaking_operator_first_num = result['shaking_method_first_num']
            local_search_oeprator_weight_list.append(local_search_oeprator_weight)
            shaking_operator_first_num_list.append(shaking_operator_first_num)
# 保存结果到CSV文件
# total_results.to_csv('collected_results.csv', index=False)
```

```python
avg_local_search_weight = {"NS1":0, "NS2":0, "NS3":0, "NS5":0, "NS6":0, "NS7":0, "NS8":0}
avg_shaking_first_num = {"NS1":0, "NS2":0, "NS6":0, "NS7":0}
for operator in avg_local_search_weight.keys():
    total_weight = 0
    total_first_num = 0
    for weight_dict in local_search_oeprator_weight_list:
        total_weight += weight_dict[operator]
    # for first_num_dict in shaking_operator_first_num_list:
    #     total_first_num += first_num_dict[operator]
    avg_local_search_weight[operator] = total_weight / len(local_search_oeprator_weight_list)
    # avg_shaking_first_num[operator] = total_first_num / len(shaking_operator_first_num_list)
for operator in avg_shaking_first_num.keys():
    total_first_num = 0
    for first_num_dict in shaking_operator_first_num_list:
        total_first_num += first_num_dict[operator]
    avg_shaking_first_num[operator] = total_first_num / len(shaking_operator_first_num_list)
```
